## 2. The question: decision, action, cost of a wrong call

### Research Question


* **Unit of Analysis:** A single content item aggregated over a historical performance window.
* **The Decision:** Prioritizing the pages for the editors to update.
* **The Action:** Updating facts or changing titles and metadata.
* **Cost of a Wrong Call:** Wasting editorial time and budget on pages that don't need fixing which can potentially risk break existing rankings.
* **Why Data / ML Helps:** ML allows us to combine non-linear decay signals across multiple dimensions and rank candidates by actionable priority.
