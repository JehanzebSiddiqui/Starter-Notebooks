## 1. My lane (or freestyle) and why

Editors don't have time to review thousands of pages. Therefore, Lane 2 builds a ranked review queue with clear reason codes so editors know exactly which high-value pages to be prioritized first.

## 2. The question: decision, action, cost of a wrong call

### Research Question
*Which high-demand, historically successful pages are experiencing non-seasonal, sustained performance decay and should be prioritized for content review first?*

* **Unit of Analysis:** A single content item (`content_hash_id`) aggregated over a historical performance window.
* **The Decision:** An SEO or content manager deciding whether to assign a page to an editor for an update, rewrite, or expansion.
* **The Action:** Updating stale facts, improving search intent alignment, expanding thin sections, or re-optimizing titles and metadata.
* **Cost of a Wrong Call:** 
  * *False Positive (recommending a healthy page):* Wasted editorial budget (~2–5 hours of writer time) and potential risk of breaking existing rankings.
  * *False Negative (missing a declining page):* Compounding loss of organic search traffic and market share to competitors over 3–6 months.
* **Why Data / ML Helps:** Simple heuristics (like `age > 180 days`) generate massive unranked lists with high noise. ML allows us to combine non-linear decay signals across multiple dimensions and rank candidates by actionable priority.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

# Load starter dataset
df = pd.read_csv('../data/raw/content_refresh_anonymized.csv')

# Filter for eligible pages
total_pages = len(df)
eligible_pages = df[(df['impressions_90d'] > 0) & (df['content_age_days'] >= 90)]
pct_eligible = (len(eligible_pages) / total_pages) * 100

# Calculate declining pages
declining_pages = eligible_pages[eligible_pages['trend_direction'] == 'down']
pct_declining = (len(declining_pages) / len(eligible_pages)) * 100

# Median impressions
median_imp_declining = declining_pages['impressions_90d'].median()
median_imp_stable = eligible_pages[eligible_pages['trend_direction'] != 'down']['impressions_90d'].median()

print(f"Total Pages: {total_pages:,} | Eligible (Age >= 90d & Imp > 0): {len(eligible_pages):,} ({pct_eligible:.1f}%)")
print(f"Declining Eligible Pages: {len(declining_pages):,} ({pct_declining:.1f}%)")
print(f"Median 90d Impressions — Declining: {median_imp_declining:,.0f} vs Stable: {median_imp_stable:,.0f}")
```
